````python
readings = torch.tensor([
    [1.2, 1.5, 1.1],
    [1.3, 1.4, 0.9],
    [1.1, 1.7, 1.0],
    [1.4, 1.6, 0.8]
])

# rows = time_steps and columns = sensors

sensor_1_avg = readings.mean(dim=0)[0].item()  # Average of sensor 1 readings
print(f"Average reading from sensor 1: {sensor_1_avg}")
min_per_sensor = readings.min(dim=0).values.tolist()  # Minimum readings per sensor
print(f"Minimum readings per sensor: {min_per_sensor}")
readings_under_1 = readings[readings < 1.0]  # All readings under 1.0
print(f"Readings under 1.0: {readings_under_1}")

readings[readings < 1.0] = 1.0  # Replace readings under 1.0 with 1.0
print(f"Updated readings: \n{readings}")

print(readings.mean().item())  # Average of all readings
````

````python
positions = torch.tensor([
    [0.0, 0.0],
    [1.0, 0.5],
    [2.0, 1.0],
    [3.0, 1.5],
    [4.0, 2.0]
])

x_tensor = positions[:, 0]
y_tensor = positions[:, 1]

dist_tensor = torch.sqrt((x_tensor[1:]-x_tensor[:-1])**2 + (y_tensor[1:]-y_tensor[:-1])**2)
print(f"Distances between consecutive positions: {dist_tensor}")
dist_tensor = dist_tensor.unsqueeze(1)  # Reshape to (4, 1)
print(f"Reshaped distances: {dist_tensor}")

print(f"Total distance traveled: {dist_tensor.sum().item():.2f}")
````
